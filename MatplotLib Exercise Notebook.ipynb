## Exercise 3 - Draw Barchart




```python
x=np.array([2,4,6,8,10])
y=np.array([5,9,7,12,15])
plt.scatter(x,y)
plt.title("Scatter Plot")
plt.show()
x=np.array(["A","B","C","D"])
y=np.array([12,7,15,10])
plt.bar(x,y)
plt.title("Bar Plot")
plt.show
# I think the scatter ploy allows us see data spreads better. The bar plot helps us see data in a more detailed manner.
```

## Exercise 4 -  Draw Histogram


```python
data=np.random.normal(50,8,200)
plt.hist(data,bins=30,density=True, color='Blue',edgecolor='black', alpha=0.7)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('Histogram')
plt.show()
# The larger the bins, the smaller they got visually.
```

## Exercise 5 -  Draw Piechart



```python
y=np.array([40,30,20,10])
plt.pie(y)
plt.title('Pie chart')
plt.show
#The largest is blue and the smallest is red. 
#The sum of all the values is 100. I think the slices reflect the percentage in relation to the whole pie. 
```
